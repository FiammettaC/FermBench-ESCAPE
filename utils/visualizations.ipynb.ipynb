```python
df = pd.read_excel("/content/dataset_curated.xlsx")
# 2. Split and clean the topics (Explode handles cells with multiple topics)
topics_series = df['topic'].dropna().str.split(',').explode().str.strip().str.title()

# 3. Complete Theme Mapping (This ensures all minor labels are grouped)
theme_mapping = {
    # Modelling & Optimization
    "Modelling": "Modelling & Optimization",
    "Experimental Design": "Modelling & Optimization",
    "Design Of Experiments": "Modelling & Optimization",

    # Bioprocess Engineering
    "Scaling": "Bioprocess Engineering",
    "Bioreactor": "Bioprocess Engineering",
    "Bioreactors Operation": "Bioprocess Engineering",
    "Continuous Fermentation": "Bioprocess Engineering",
    "Oxygen Transfer": "Bioprocess Engineering",
    "Bioprocess Development": "Bioprocess Engineering",
    "Industrial Biotech": "Bioprocess Engineering",

    # Organisms & Cell Biology
    "Bacterial": "Organisms & Cell Biology",
    "Yeast": "Organisms & Cell Biology",
    "Mold": "Organisms & Cell Biology",
    "Pure Strain": "Organisms & Cell Biology",
    "Cell Physiology": "Organisms & Cell Biology",
    "Biology": "Organisms & Cell Biology",

    # Sector Applications
    "Food": "Sector Applications",
    "Food Biotech": "Sector Applications",
    "Pharma": "Sector Applications",
    "Med Biotech": "Sector Applications",
    "Biofuel": "Sector Applications",
    "Alcohol": "Sector Applications",
    "Bioplastic": "Sector Applications",
    "Waste Treatment": "Sector Applications",

    # Specific Biochemistry & Operations
    "Enzymes": "Biochemistry & Operations",
    "Medium Composition": "Biochemistry & Operations",
    "Mixed Media": "Biochemistry & Operations",
    "Yield": "Biochemistry & Operations",
    "Product Toxicity": "Biochemistry & Operations",

    # Fundamentals
    "Basics": "Fundamentals"
}

# 4. Apply the mapping
themed_topics = topics_series.replace(theme_mapping)
theme_counts = themed_topics.value_counts()

# 5. Generate the Plot
plt.figure(figsize=(12, 7))

# Apply a colorful palette and sort by count (value_counts does this by default)
colors = plt.cm.Paired(range(len(theme_counts)))
plt.bar(theme_counts.index, theme_counts.values, color=colors, edgecolor='black')

# Styling
plt.xticks(rotation=35, ha='right', fontsize=15)
plt.ylabel('Number of Questions', fontsize=15)
#plt.title('Distribution of Fermentation Questions by Theme', fontsize=15, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

# Save and Show
plt.savefig('thematic_distribution_complete.png')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates

data_gpt = {
    'Chatbot': ['ChatGPT', 'Claude', 'Gemini', 'DeepSeek', 'Le Chat'],
    'Factual accuracy': [4.92, 4.52, 4.33, 4.59, 4.32],
    'Comprehensive': [4.94, 4.29, 4.88, 4.94, 4.71],
    'Coherence': [4.98, 4.73, 4.96, 4.97, 4.83],
    'Concision': [3.38, 3.86, 3.32, 3.13, 3.47],
    'Readability': [4.94, 4.79, 4.97, 4.98, 4.97],
    'Helpfulness': [4.94, 4.54, 4.90, 4.96, 4.67],
    'Insightfulness': [4.06, 3.59, 4.06, 4.02, 3.72],
    'Overall quality': [4.94, 4.41, 4.64, 4.77, 4.55]
}

df = pd.DataFrame(data_gpt)
distinct_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# 2. Create the plot
plt.figure(figsize=(12, 6))
# 'Chatbot' is the column used to color the lines
parallel_coordinates(df, 'Chatbot', color=distinct_colors)

# 3. Styling and Labels
plt.title('Chatbot Performance According to GPT 5.2', fontsize=14)
plt.ylabel('Score (0-5)', fontsize=12)
plt.legend(title='Chatbots', loc='upper right', bbox_to_anchor=(1.15, 1.0))
plt.grid(True, alpha=0.8)
plt.tight_layout()
#plt.ylim(0, 6)

# 4. Save and Show
plt.savefig('parallel_coordinates_plot_gpt.png')
plt.show()
```
